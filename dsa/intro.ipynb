```python
card=[1,0,12.4,3,7,13,11]
qury=7
```

```python
def locate_cards(card, qury):
    for position in (0, len(card)):
        
        while True:
            if card[position] == qury:
                return position
            position += 1
            if position == len(card):
                return -1
        print(position)
```

```python
result = locate_cards(card, qury)
print("The value is at index:", result)
```

```python
def locate_cards(card, qury):
    # Iterate through all valid indices
    for position in range(len(card)):
        # Check if the current element matches the query
        if card[position] == qury:
            return position  
            
    # If the loop finishes without finding the query
    return -1
```

```python
result = locate_cards(card, qury)
print("The value is at index:", result)
```
